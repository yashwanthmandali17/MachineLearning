```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import make_classification
```

```python
x,y=make_classification(n_samples=1000,n_features=2,n_informative=2,n_redundant=0,n_clusters_per_class=2)
```

```python
x
```

```python
y
```

```python
pd.DataFrame(x[0])
```

```python
sns.scatterplot(
    x=pd.DataFrame(x)[0],
    y=pd.DataFrame(x)[1],
    hue=y
)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=10)
```

```python
from sklearn.svm import SVC
svc=SVC(kernel='linear')
svc.fit(x_train,y_train)
```

```python
svc.coef_
```

```python
y_pred=svc.predict(x_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
print('Classification_report:\n',classification_report(y_test,y_pred))
print('confusion_matrix:\n',confusion_matrix(y_test,y_pred))
print('accuracy_score:\n',accuracy_score(y_test,y_pred))
```

```python
rbf=SVC(kernel='rbf')
rbf.fit(x_train,y_train)
```

```python
y_pred1=rbf.predict(x_test)
```

```python
print('Classification_report:\n',classification_report(y_test,y_pred1))
print('confusion_matrix:\n',confusion_matrix(y_test,y_pred1))
print('accuracy_score:\n',accuracy_score(y_test,y_pred1))
```

```python
pol=SVC(kernel='poly')
pol.fit(x_train,y_train)
```

```python
y_pred2=pol.predict(x_test)
```

```python
print('Classification_report:\n',classification_report(y_test,y_pred2))
print('confusion_matrix:\n',confusion_matrix(y_test,y_pred2))
print('accuracy_score:\n',accuracy_score(y_test,y_pred2))
```

```python
s=SVC(kernel='sigmoid')
s.fit(x_train,y_train)
```

```python
y_pred3=s.predict(x_test)
```

```python
print('Classification_report:\n',classification_report(y_test,y_pred3))
print('confusion_matrix:\n',confusion_matrix(y_test,y_pred3))
print('accuracy_score:\n',accuracy_score(y_test,y_pred3))
```

### Accuracies
1. linear ->93

2. rbf ->95

4. sigmoid ->78

```python
from sklearn.model_selection import GridSearchCV
```

```python
pram_grid={'C':[0.1,1,10,100,1000],
           'gamma':[1,0.1,0.01,0.001,0.0001],
           'kernel':['rbf']}
```

```python
grid=GridSearchCV(SVC(),param_grid=pram_grid,refit=True,cv=5,verbose=3)
```

```python
grid.fit(x_train,y_train)
```

```python
grid.best_params_
```

```python
y_pred4=grid.predict(x_test)
print('Classification_report:\n',classification_report(y_test,y_pred4))
print('confusion_matrix:\n',confusion_matrix(y_test,y_pred4))
print('accuracy_score:\n',accuracy_score(y_test,y_pred4))
```

C=Regularization Parameter

gamma=Influence of a single data point

total combinations:
5(c)*5(gamma)*1(kernel)=25 models
